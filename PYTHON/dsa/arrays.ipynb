```python
#3 sum


a = [1,0,-1,2,-1,4]
n = len(a)

s = set()
for i in range(n):
    for j in range(i+1,n):
        for k in range(j+1,n):

            if a[i]+a[j]+a[k]==0:

                t = tuple(([a[i],a[j],a[k]]))

                if t not in s:
                    print(t)
                    s.add(t)
      
            

```
