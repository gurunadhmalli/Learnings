```python
#loop through dict

a = {"name":"guru","age":21}

for i in a.keys():
    print(i)

for i in a.values():
    print(i)

for i,j in a.items():
    print(i,j)
```

```python
#copy

#this is the problem this is not copying and this jsut referencing
a = {"name":"guru","age":21}
print(id(a))
b = a
print(id(b))
a["rool_no"] = 221
print(b)


#we have the built in method copy
print("copy")
a = {"name":"guru","age":21}
print(id(a))
b = a.copy()
print(id(b))
a["rool_no"] = 221
print(a)
print(b)


#we have a built in method dict
print("dict")
a = {"name":"guru","age":21}
print(id(a))
b = dict(a)
print(id(b))
a["rool_no"] = 221
print(a)
print(b)
```
