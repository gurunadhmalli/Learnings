```python
#strign creation 

import sys
a = "guru"
b = 'guru'
print(a)
print(b)
print(id(a))
print(id(b))

#get the size of the string

print(sys.getsizeof(a))
print(sys.getsizeof(b))



#get the reference count

print(sys.getrefcount(a))
print(sys.getrefcount(b))
```

```python
#quotes inside a quotes


a = "it's okay"
b = "my name is 'guru'"
c = 'my name is "guru"'
print(a)
print(b)
print(c)
```

```python
#multi line string 

a = '''my name is gurunadh 
i am 22 years old and i am currently doing
a job as ai engineer'''

b = """'''my name is gurunadh 
i am 22 years old and i am currently doing
a job as ai engineer'''"""

print(a)
print(b)
print(id(a))
print(id(b))
```

```python
#looping through a string 


a = "guru"

for i in a:
    print(i)
    print(id(i))
    print(type(i))
    
```
