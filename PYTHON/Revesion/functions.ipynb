```python
#function with argument 
#with one argument 

def greet(name):
    print("hello",name)

greet("guru")



#with multiple parametres

def intro(name,age,city):
    print(name)
    print(age)
    print(city)

intro("gurunadh",21,"kota")
```

```python
#types of the argurments


#default argument

def greet(name="guru",message="good morning"):
    print(message,name)

greet("srinu")
```

```python
#postional arguments

def add(a,b):
    return a+b

result=add(10,10)
print(result)


#order matters
def sub(a,b):
    return a-b

print(sub(5,10))
```

```python
#keyword argument

#order does not matter
def sub(a,b):
    return a-b

res = sub(b=5,a = 10)
print(res)
```

```python
# we can combine this default + positional + keyword


def intro(age,city,name="guruandh"):
    return f"{name},{age},{city}"


print(intro(21,city="kota"))
```

```python
#variable lenght positional arguments

def add(*num):
    print(num)

add(1,2,3,4,5,6,7)


def add(*args):
    return sum(args)

print(add(1,2,3))

```

```python
#varible length keyword arguments 


def show_details(**details):
    print(details)

show_details(name= "guru",age=21,city="kota")
```

```python
#combunation of the args and kwargs


def test(*args,**kwargs):
    print(args)
    print(kwargs)


test(1,2,3,4,5,name= "guru",age = 21,city = "kota")
```

```python
#argument unpacking


a = {10,20,30}

def add(a,b,c):
    return a+b+c

print(add(*a))


d = {"name":"guru","age":22}
def show_details(name,age,city="kota"):
    print(name,age,city)

show_details(**d)


```
