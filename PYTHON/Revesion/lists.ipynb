```python
a = [10,20]
print(a)
print("this is the list id",id(a))
print("id of the list a")
for i in a:
    print(id(i))



b = [30,40]
print(b)
print("id of the list b")
for i in b:
    print(id(i))



a.extend(b)
print("this is the extended list id",id(a))
print(a)
print("extended list")
for i in a:
    print(id(i))

```
