```python
#many values to many vaaraibles
a,b,c = 10,20,30
print(a,b,c)
print(id(a))
print(id(b))
print(id(c))
```

```python
#one valaue to many varaibles
a = b = c = 10
print(a)
print(id(a))
print(id(b))
print(id(c))
```
