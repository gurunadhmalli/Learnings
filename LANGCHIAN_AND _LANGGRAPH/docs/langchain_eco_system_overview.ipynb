```text
                         +-----------------------+
                         |       LangChain       |
                         |-----------------------|
                         | Core framework for    |
                         | building LLM apps.    |
                         | Provides chains,      |
                         | prompts, agents,      |
                         | RAG components, etc.  |
                         +----------+------------+
                                    |
        +---------------------------+----------------------------+
        |                            |                            |
        v                            v                            v
+------------------+       +------------------+        +------------------+
|    LangGraph     |       |    LangSmith     |        |    LangServe     |
|------------------|       |------------------|        |------------------|
| Stateful agent   |       | Tracing,         |        | Deploy LangChain |
| orchestration.   |       | monitoring,      |        | apps as REST     |
| Builds complex   |       | debugging, and   |        | APIs using       |
| multi-agent and  |       | evaluation of    |        | FastAPI.          |
| cyclic workflows.|       | LLM applications.|        |                  |
+------------------+       +------------------+        +------------------+
                                    |
                                    |
                    +---------------+----------------+
                    |                                |
                    v                                v
          +----------------------+         +----------------------+
          |  Integration Packages|         | Community Packages   |
          |----------------------|         |----------------------|
          | langchain-openai     |         | langchain-community  |
          | langchain-anthropic  |         | Document loaders,    |
          | langchain-google     |         | vector stores,       |
          | langchain-cohere     |         | tools, utilities,    |
          | Connects to LLMs     |         | and third-party      |
          | and embedding APIs.  |         | integrations.        |
          +----------------------+         +----------------------+
```

## Components

### 1. `langchain-core`
- The foundation of the ecosystem.
- Contains **base abstractions** such as:
  - Prompt Templates
  - Runnable interfaces
  - Messages
  - Output Parsers
  - LCEL (LangChain Expression Language)

> **Think of it as the engine and common building blocks.**

---

### 2. `langchain`
- The main high-level framework.
- Used to build:
  - Chains
  - Agents
  - RAG pipelines
  - Tool calling workflows

> **Think of it as the application builder.**

---

### 3. `langgraph`
- Built on top of LangChain.
- Used for creating **stateful AI agents** and **multi-step workflows**.
- Supports:
  - Agent memory/state
  - Loops and retries
  - Multi-agent collaboration
  - Human-in-the-loop systems

> **Think of it as the workflow orchestrator or brain.**

Example:
```
User Query
    |
    v
+-----------+
|  Planner  |
+-----------+
    |
+---+---+
|       |
v       v
Search  SQL DB
  |       |
  +---+---+
      |
      v
+-----------+
|  LLM Node |
+-----------+
      |
      v
Final Answer
```

---

### 4. `langsmith`
- Developer platform for:
  - Tracing
  - Debugging
  - Monitoring
  - Evaluating LLM applications

It records:
- User input
- Prompt sent to the LLM
- Retrieved documents
- Tool calls
- LLM outputs
- Latency and token usage

> **Think of it as the debugger and analytics dashboard.**

---

### 5. `langserve`
- Deploys LangChain/LangGraph applications as production-ready REST APIs.
- Built on top of FastAPI.

Example:
```
POST /chat
{
    "question": "What is RAG?"
}

Response:
{
    "answer": "RAG stands for Retrieval-Augmented Generation..."
}
```

> **Think of it as the deployment layer.**

---

## Integration Packages

| Package | Purpose |
|----------|----------|
| `langchain-openai` | Connects to OpenAI models and embeddings |
| `langchain-anthropic` | Connects to Anthropic Claude models |
| `langchain-google-genai` | Connects to Google Gemini models |
| `langchain-cohere` | Connects to Cohere models |
| `langchain-huggingface` | Connects to Hugging Face models |

> These packages allow LangChain to communicate with different LLM providers.

---

## Community Package

### `langchain-community`
Contains community-maintained integrations such as:
- PDF Loaders
- CSV Loaders
- Web Loaders
- FAISS
- Chroma
- Pinecone
- SQL Databases
- Search tools
- Utility functions

> **Think of it as the plugin library for LangChain.**

# Complete Flow

```text
                    +------------------+
                    |      User        |
                    +---------+--------+
                              |
                              v
                    +---------+--------+
                    |    LangGraph     |
                    | (Agent Workflow) |
                    +---------+--------+
                              |
                              v
                    +---------+--------+
                    |    LangChain     |
                    | (Chains, Agents, |
                    | Prompts, RAG)    |
                    +---------+--------+
                              |
        +---------------------+----------------------+
        |                     |                      |
        v                     v                      v
+---------------+   +----------------+   +------------------+
| LLM Providers |   | Vector Database|   | External Tools   |
| OpenAI        |   | FAISS, Chroma  |   | Search, SQL, API |
| Claude        |   | Pinecone       |   | Python, etc.     |
| Gemini        |   +----------------+   +------------------+
| Cohere        |
+---------------+
        |
        v
+----------------------+
|      LangSmith       |
|  Trace • Monitor •   |
|  Evaluate • Debug    |
+----------------------+
        |
        v
+----------------------+
|      LangServe       |
| Deploy as REST APIs  |
+----------------------+
```
